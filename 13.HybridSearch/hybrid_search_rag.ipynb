```python
from pinecone import Pinecone,ServerlessSpec
import os
from dotenv import load_dotenv

load_dotenv()
```

```python
# Creating a Pinecone Index Through Code
api_key=os.environ['PINECONE_API_KEY']
print(api_key)
index_name="hybrid-search-langchain-learn"

client=Pinecone(api_key=api_key)

if index_name not in client.list_indexes().names():
    index=client.create_index(
        name=index_name,
        dimension=384,
        metric='dotproduct',
        spec=ServerlessSpec(cloud='aws',region='us-east-1'),
    )
    print('Index Created')
```

```python
# connect to index

index=client.Index(index_name)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.retrievers import PineconeHybridSearchRetriever
from pinecone_text.sparse import BM25Encoder

embeddings_model=HuggingFaceEmbeddings(model_name='sentence-transformers/all-MiniLM-L6-v2')
encoder=BM25Encoder().default()

retriever=PineconeHybridSearchRetriever(
    name=index_name,
    index=index,
    embeddings=embeddings_model,
    sparse_encoder=encoder
)
```

```python
texts = [
    "Artificial intelligence is transforming industries by enabling machines to learn from data.",
    "Machine learning models improve their performance as they are exposed to more training data.",
    "LangChain is a framework used to build applications powered by large language models.",
    "Vector databases like Pinecone are used to store and retrieve embeddings efficiently.",
    "Hybrid search combines dense vector search with traditional keyword-based search.",
    "Python is a popular programming language for data science and machine learning.",
    "Transformers are deep learning models that are widely used in natural language processing.",
    "Retrieval-Augmented Generation improves response accuracy by fetching relevant documents.",
    "The MiniLM model is a lightweight transformer model used for generating embeddings.",
    "FAISS is a library developed by Meta for efficient similarity search and clustering.",
    "Natural language processing enables computers to understand and generate human language.",
    "APIs allow different software systems to communicate with each other.",
    "Cloud platforms like AWS provide scalable infrastructure for deploying AI applications.",
    "Embedding models convert text into numerical vectors that capture semantic meaning.",
    "Chatbots use language models to interact with users in a conversational manner."
]

retriever.add_texts(texts=texts)
```

```python
query = "What is hybrid search?"

response=retriever.invoke(input=query)

print(response)
```
